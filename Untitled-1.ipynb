```python
import os
import cv2
import numpy as np
from skimage.feature import local_binary_pattern
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
radius = 1
n_points = 8 * radius
```

```python
def extract_lbp(image):
    lbp = local_binary_pattern(image, n_points, radius, method="uniform")
    hist, _ = np.histogram(
        lbp.ravel(),
        bins=np.arange(0, n_points + 3),
        range=(0, n_points + 2)
    )
    hist = hist.astype("float")
    hist /= (hist.sum() + 1e-6)
    return hist
```

```python
def load_dataset(path):
    X = []
    y = []

    for label in os.listdir(path):
        folder = os.path.join(path, label)

        if not os.path.isdir(folder):
            continue

        for file in os.listdir(folder):
            img_path = os.path.join(folder, file)

            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is None:
                continue

            img = cv2.resize(img, (48, 48))
            features = extract_lbp(img)

            X.append(features)
            y.append(label)

    return np.array(X), np.array(y)
```

```python
X_train, y_train = load_dataset("C:/Users/user/Desktop/TD2/archive/train")
X_test, y_test = load_dataset("C:/Users/user/Desktop/TD2/archive/test")
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

print("Modèle entraîné ✅")
```

```python
y_pred = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
def extract_hog(image):
    features = hog(
        image,
        orientations=9,
        pixels_per_cell=(8, 8),
        cells_per_block=(2, 2),
        block_norm='L2-Hys'
    )
    return features
```

```python
def load_dataset_hog(path):
    X = []
    y = []

    for label in os.listdir(path):
        folder = os.path.join(path, label)

        if not os.path.isdir(folder):
            continue

        for file in os.listdir(folder):  # ❌ enlever [:500]
            img_path = os.path.join(folder, file)

            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is None:
                continue

            img = cv2.resize(img, (48, 48))
            features = extract_hog(img)

            X.append(features)
            y.append(label)

    return np.array(X), np.array(y)
```
